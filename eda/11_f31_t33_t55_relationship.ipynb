# 11. F31, T33 и T55: связь расхода и температур

`F31` — расход нефтепродукта в П-3; `T33` — температура низа К-2; `T55` — температура на выходе печи П-3. В расчёте разделены уровни, 10-минутные изменения и временные лаги. Корреляция не доказывает причинное воздействие и не является рекомендацией по управлению.

```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import display, IFrame

HERE = Path.cwd().resolve()
EDA_DIR = HERE if HERE.name == 'eda' else HERE / 'eda'
sys.path.insert(0, str(EDA_DIR))
%run $EDA_DIR/analyze_f31_t33_t55.py
OUT = EDA_DIR / 'artifacts' / 'f31_t33_t55'
```

```python
display(pd.read_csv(OUT / 'correlations.csv'))
display(pd.read_csv(OUT / 'yearly_running_level_correlations.csv'))
```
